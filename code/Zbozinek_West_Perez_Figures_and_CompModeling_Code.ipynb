```python
# Figure 4

# Load and filter the data
df = pd.read_csv(latent_learning_path + 'OS_Transfer_Latent_Learning.csv')
df = df[df['stimulus_index'].isin([1, 2, 3])]
df = df[df['TransferTraining'].isin([1, 2, 3])]

# Set up figure
p = figure(
    frame_width=700,
    frame_height=400,
    x_axis_label='Hierarchical Training',
    y_axis_label='Agreement',
    x_range=(-0.5, 1.5),
    y_range=(0, 125),
    toolbar_location=None,
    tools='',
)

# Plotting parameters
bar_width = 0.25
scatter_size = 4
line_alpha = 0.4

# Color mapping for stimulus_index
stim_colors = {1: Color1, 2: Color2, 3: Color3}
stim_labels = {1: '50% PR', 2: 'Modulation', 3: 'Configuration'}

# X-axis labels
x_labels = {0: 'PR Control', 1: 'True OS'}

# Store bar positions and summary values for annotations
bar_positions = {}
bar_summaries = {}

# Add dummy bars for legend
for stim_idx in [1, 2, 3]:
    p.vbar(
        x=-1,
        top=0,
        width=0.01,
        color=stim_colors[stim_idx],
        alpha=0.4,
        legend_label=stim_labels[stim_idx],
    )

# Iterate over HierarchicalTraining and stimulus_index
for ht_val in [0, 1]:
    for stim_idx in [1, 2, 3]:
        group_df = df[
            (df['HierarchicalTraining'] == ht_val)
            & (df['stimulus_index'] == stim_idx)
        ].sort_values('DV').reset_index(drop=True)

        n = len(group_df)
        if n == 0:
            continue

        color = stim_colors[stim_idx]
        x_base = ht_val + (-0.25 + (stim_idx - 1) * bar_width)

        bar_positions[(ht_val, stim_idx)] = x_base

        # ECDF-style offset
        x_offsets = (
            np.linspace(-bar_width / 2, bar_width / 2, n)
            if n > 1 else np.array([0.0])
        )

        for idx, row in group_df.iterrows():
            x = x_base + x_offsets[idx]
            y = row['DV']

            p.scatter(
                x=[x],
                y=[y],
                size=scatter_size,
                alpha=0.8,
                color=color,
            )

        # Mean and error bar
        mean = group_df['DV'].mean()
        sem = group_df['DV'].sem()

        bar_summaries[(ht_val, stim_idx)] = {
            'mean': mean,
            'sem': 0 if pd.isna(sem) else sem,
            'max': group_df['DV'].max(),
        }

        p.vbar(
            x=x_base,
            top=mean,
            width=bar_width,
            color=color,
            alpha=0.4,
        )

        p.segment(
            x0=x_base,
            y0=mean - sem,
            x1=x_base,
            y1=mean + sem,
            line_color='gray',
            line_width=1,
            alpha=0.7,
        )

        p.segment(
            x0=x_base - 0.05,
            y0=mean + sem,
            x1=x_base + 0.05,
            y1=mean + sem,
            line_color='gray',
            line_width=1,
            alpha=0.7,
        )

        p.segment(
            x0=x_base - 0.05,
            y0=mean - sem,
            x1=x_base + 0.05,
            y1=mean - sem,
            line_color='gray',
            line_width=1,
            alpha=0.7,
        )

# Connect each ID's points across stimulus_index
for ht_val in [0, 1]:
    ht_df = df[df['HierarchicalTraining'] == ht_val]

    for id_val in ht_df['ID'].unique():
        x_vals = []
        y_vals = []

        for stim_idx in [1, 2, 3]:
            stim_df = ht_df[
                ht_df['stimulus_index'] == stim_idx
            ].sort_values('DV').reset_index(drop=True)

            if id_val not in stim_df['ID'].values:
                continue

            match_idx = stim_df[stim_df['ID'] == id_val].index[0]
            x_base = ht_val + (-0.25 + (stim_idx - 1) * bar_width)

            x_offsets = (
                np.linspace(-bar_width / 2, bar_width / 2, len(stim_df))
                if len(stim_df) > 1 else np.array([0.0])
            )

            x_vals.append(x_base + x_offsets[match_idx])
            y_vals.append(stim_df.loc[match_idx, 'DV'])

        if len(x_vals) >= 2:
            p.line(
                x=x_vals,
                y=y_vals,
                line_color='gray',
                line_width=0.75,
                alpha=0.2,
            )

# --------------------------------------------------
# Add significance-style brackets and asterisks
# --------------------------------------------------
# These annotations are fixed as requested; no statistical
# tests are performed by this code.


comparisons = [
    ((0, 1), (1, 1), '*'),
    ((0, 2), (1, 2), '***'),
    ((0, 3), (1, 3), '***'),
    ((1, 1), (1, 2), '***'),
    ((1, 1), (1, 3), '***'),
]

# Determine the highest observed/error-bar value for each comparison
bracket_specs = []

for order, (left_key, right_key, stars) in enumerate(comparisons):
    if left_key not in bar_positions or right_key not in bar_positions:
        print(
            f"Skipping annotation {stars}: "
            f"missing bar for {left_key} or {right_key}."
        )
        continue

    left_summary = bar_summaries[left_key]
    right_summary = bar_summaries[right_key]

    highest_value = max(
        left_summary['max'],
        right_summary['max'],
        left_summary['mean'] + left_summary['sem'],
        right_summary['mean'] + right_summary['sem'],
    )

    bracket_specs.append({
        'left_x': bar_positions[left_key],
        'right_x': bar_positions[right_key],
        'stars': stars,
        'height': highest_value,
        'order': order,
    })

# Place brackets in the requested bottom-to-top order
bracket_specs.sort(key=lambda item: item['order'])

placed_brackets = []
bracket_gap = 5
bracket_height = 2

for spec in bracket_specs:
    left_x = spec['left_x']
    right_x = spec['right_x']

    # Start at a common baseline, then move upward to avoid overlap
    base_y = 106 + spec['order'] * 8

    for placed in placed_brackets:
        overlaps = (
            left_x <= placed['right_x']
            and right_x >= placed['left_x']
        )

        if overlaps and base_y <= placed['top_y'] + bracket_gap:
            base_y = placed['top_y'] + bracket_gap

    top_y = base_y + bracket_height

    p.segment(
        x0=left_x, y0=top_y,
        x1=right_x, y1=top_y,
        line_color='black', line_width=1.5,
    )

    p.segment(
        x0=left_x, y0=base_y,
        x1=left_x, y1=top_y,
        line_color='black', line_width=1.5,
    )

    p.segment(
        x0=right_x, y0=base_y,
        x1=right_x, y1=top_y,
        line_color='black', line_width=1.5,
    )

    p.text(
        x=[(left_x + right_x) / 2],
        y=[top_y + 0.5],
        text=[spec['stars']],
        text_align='center',
        text_baseline='bottom',
        text_font_size='14pt',
        text_font_style='bold',
        text_color='black',
    )

    placed_brackets.append({
        'left_x': left_x,
        'right_x': right_x,
        'top_y': top_y + 5,
    })

if placed_brackets:
    p.y_range.end = max(
        125,
        max(item['top_y'] for item in placed_brackets) + 5
    )


# Axis customization
p.xaxis.ticker = [0, 1]
p.xaxis.major_label_overrides = x_labels

p.yaxis.ticker = [0, 50, 100]
p.yaxis.major_label_overrides = {
    0: 'Strongly Disagree',
    50: 'Neither Agree\nNor Disagree',
    100: 'Strongly Agree',
}

p.legend.location = 'top_left'
p.legend.orientation = 'vertical'
p.legend.click_policy = 'hide'
p.legend.label_text_font_size = LegendSize

p.xaxis.axis_label_text_font_size = AxisLabelSize
p.xaxis.major_label_text_font_size = TickSize
p.yaxis.axis_label_text_font_size = AxisLabelSize
p.yaxis.major_label_text_font_size = TickSize
p.yaxis.axis_label_text_font_style = 'normal'
p.xaxis.axis_label_text_font_style = 'normal'


# Show and export
show(p)

with contextlib.redirect_stdout(io.StringIO()):
    export_png(
        p,
        filename=os.path.join(figures_path, 'Figure 4.png')
    )

```

```python
# Figure 5

# Load and filter the data
df = pd.read_csv(os.path.join(exp_test_path, 'OS_Transfer_Exp_Test.csv'))
df = df[df['Phase'] == 'Test1']
df = df[df['TransferTraining'].isin([1, 2, 3])]

# Training group stimulus mappings
group_stims = {
    1: {'Trained': ['Ax', 'Ay', 'Bx', 'By'], 'Transfer': ['Az', 'Bz', 'Cx', 'Cy']},
    2: {'Trained': ['Ax', 'By', 'Cz'], 'Transfer': ['Ay', 'Az', 'Bx', 'Bz', 'Cx', 'Cy']},
    3: {'Trained': ['Ax', 'By'], 'Transfer': ['Az', 'Bz', 'Cx', 'Cy']}
}

# Color mapping
bar_colors = {'Trained': Color1, 'Transfer': Color5}

# Plotting parameters
bar_width = 0.3
scatter_size = 4
line_alpha = 0.4

# Plot True OS first, then PR Control
plots = []

for ht in [1, 0]:

    df_ht = df[df['HierarchicalTraining'] == ht]

    p = bokeh.plotting.figure(
        frame_width=700,
        frame_height=400,
        title='True OS' if ht == 1 else 'PR Control',
        x_axis_label='Transfer Training',
        y_axis_label='US Expectancy',
        x_range=(-0.25, 3.5),
        y_range=(0, 115),
        toolbar_location=None,
        tools='',
    )

    for i, trans in enumerate([1, 2, 3], start=1):

        pair = group_stims[trans]
        x_trained = i - 0.2
        x_transfer = i + 0.2

        group_df = df_ht[df_ht['TransferTraining'] == trans]

        # Compute per-ID averages
        id_data = []

        for id_val, df_id in group_df.groupby('ID'):
            entry = {'ID': id_val}

            for label in ['Trained', 'Transfer']:
                subset = df_id[df_id['Procedure'].isin(pair[label])]
                entry[label] = subset['DV'].mean() if not subset.empty else np.nan

            id_data.append(entry)

        id_df = pd.DataFrame(id_data).dropna(
            subset=['Trained', 'Transfer']
        )

        if id_df.empty:
            continue

        # Sort values independently for scatter positioning
        sorted_trained_df = id_df.sort_values(
            by='Trained'
        ).reset_index(drop=True)

        sorted_transfer_df = id_df.sort_values(
            by='Transfer'
        ).reset_index(drop=True)

        n = len(id_df)

        x_offsets = (
            np.linspace(-bar_width / 2, bar_width / 2, n)
            if n > 1 else np.array([0.0])
        )

        # Plot Trained observations
        for idx, row in sorted_trained_df.iterrows():
            p.scatter(
                x=[x_trained + x_offsets[idx]],
                y=[row['Trained']],
                size=scatter_size,
                alpha=0.8,
                color=bar_colors['Trained'],
            )

        # Plot Transfer observations
        for idx, row in sorted_transfer_df.iterrows():
            p.scatter(
                x=[x_transfer + x_offsets[idx]],
                y=[row['Transfer']],
                size=scatter_size,
                alpha=0.8,
                color=bar_colors['Transfer'],
            )

        # Connect paired values by ID
        trained_positions = {
            id_val: idx
            for idx, id_val in enumerate(sorted_trained_df['ID'])
        }
        transfer_positions = {
            id_val: idx
            for idx, id_val in enumerate(sorted_transfer_df['ID'])
        }

        for _, row in id_df.iterrows():
            id_val = row['ID']

            x1 = x_trained + x_offsets[trained_positions[id_val]]
            x2 = x_transfer + x_offsets[transfer_positions[id_val]]

            p.line(
                x=[x1, x2],
                y=[row['Trained'], row['Transfer']],
                color='gray',
                line_width=1,
                alpha=line_alpha,
            )

        # Plot means and SEMs
        means = {}
        sems = {}

        for label, x_val in zip(
            ['Trained', 'Transfer'],
            [x_trained, x_transfer]
        ):
            mean_val = id_df[label].mean()
            sem_val = id_df[label].sem()

            means[label] = mean_val
            sems[label] = sem_val

            p.vbar(
                x=x_val,
                top=mean_val,
                width=bar_width,
                color=bar_colors[label],
                alpha=0.4,
                legend_label=label,
            )

            cap = 0.05

            p.segment(
                x0=x_val,
                y0=mean_val - sem_val,
                x1=x_val,
                y1=mean_val + sem_val,
                line_color='gray',
                line_width=1,
                alpha=0.7,
            )

            p.segment(
                x0=x_val - cap,
                y0=mean_val + sem_val,
                x1=x_val + cap,
                y1=mean_val + sem_val,
                line_color='gray',
                line_width=1,
                alpha=0.7,
            )

            p.segment(
                x0=x_val - cap,
                y0=mean_val - sem_val,
                x1=x_val + cap,
                y1=mean_val - sem_val,
                line_color='gray',
                line_width=1,
                alpha=0.7,
            )

        # Add dagger only for True OS, Ambiguous condition (TransferTraining = 2)
        if ht == 1 and trans == 2:

            highest_value = max(
                id_df['Trained'].max(),
                id_df['Transfer'].max(),
                means['Trained'] + sems['Trained'],
                means['Transfer'] + sems['Transfer'],
            )

            bracket_y = highest_value + 3
            bracket_height = 2

            # Horizontal bracket
            p.segment(
                x0=x_trained,
                y0=bracket_y + bracket_height,
                x1=x_transfer,
                y1=bracket_y + bracket_height,
                line_color='black',
                line_width=1.5,
            )

            # Vertical bracket ends
            p.segment(
                x0=x_trained,
                y0=bracket_y,
                x1=x_trained,
                y1=bracket_y + bracket_height,
                line_color='black',
                line_width=1.5,
            )

            p.segment(
                x0=x_transfer,
                y0=bracket_y,
                x1=x_transfer,
                y1=bracket_y + bracket_height,
                line_color='black',
                line_width=1.5,
            )

            # Dagger
            p.text(
                x=[(x_trained + x_transfer) / 2],
                y=[bracket_y + bracket_height + 0.5],
                text=['†'],
                text_align='center',
                text_baseline='bottom',
                text_font_size='14pt',
                text_font_style='bold',
                text_color='black',
            )

        
        # Add two asterisks for True OS, Specificity condition
        if ht == 1 and trans == 3:

            highest_value = max(
                id_df['Trained'].max(),
                id_df['Transfer'].max(),
                means['Trained'] + sems['Trained'],
                means['Transfer'] + sems['Transfer'],
            )

            bracket_y = highest_value + 3
            bracket_height = 2

            # Horizontal bracket
            p.segment(
                x0=x_trained,
                y0=bracket_y + bracket_height,
                x1=x_transfer,
                y1=bracket_y + bracket_height,
                line_color='black',
                line_width=1.5,
            )

            # Vertical bracket ends
            p.segment(
                x0=x_trained,
                y0=bracket_y,
                x1=x_trained,
                y1=bracket_y + bracket_height,
                line_color='black',
                line_width=1.5,
            )

            p.segment(
                x0=x_transfer,
                y0=bracket_y,
                x1=x_transfer,
                y1=bracket_y + bracket_height,
                line_color='black',
                line_width=1.5,
            )

            # Two asterisks
            p.text(
                x=[(x_trained + x_transfer) / 2],
                y=[bracket_y + bracket_height + 0.5],
                text=['**'],
                text_align='center',
                text_baseline='bottom',
                text_font_size='14pt',
                text_font_style='bold',
                text_color='black',
            )

    # Customize axes
    p.xaxis.ticker = [1, 2, 3]
    p.xaxis.major_label_overrides = {
        1: 'Generalization',
        2: 'Ambiguous',
        3: 'Specificity',
    }

    p.yaxis.ticker = [0, 50, 100]
    p.yaxis.major_label_overrides = {
        0: 'Certain "No" Money',
        50: 'Completely Uncertain',
        100: 'Certain "Yes" Money',
    }

    p.legend.location = 'center_left'
    p.legend.orientation = 'vertical'
    p.legend.click_policy = 'hide'
    p.legend.label_text_font_size = LegendSize

    p.xaxis.axis_label_text_font_size = AxisLabelSize
    p.xaxis.major_label_text_font_size = TickSize
    p.yaxis.axis_label_text_font_size = AxisLabelSize
    p.yaxis.major_label_text_font_size = TickSize

    p.yaxis.axis_label_text_font_style = 'normal'
    p.xaxis.axis_label_text_font_style = 'normal'
    p.title.text_font_size = TitleSize

    plots.append(p)

    
# Export both plots as one PNG, with True OS on top and PR Control below

combined_plot = bokeh.layouts.gridplot(
    [
        [plots[0]],
        [plots[1]],
    ],
    toolbar_location=None,
    merge_tools=False,
)

export_png(
    combined_plot,
    filename=os.path.join(
        figures_path,
        "Figure 5.png",
    ),
)

bokeh.io.show(combined_plot)
```

```python
# Computational Modeling

# Read in data and sort
df = pd.read_csv(os.path.join('../data/Exp_Ext/', 'OS_Transfer_Exp.csv')).dropna()
df['trialseq'] = df['trialseq'].astype(int)
df['US'] = df['US'].astype(int)
df['noUS'] = df['noUS'].astype(int)


# Ensure properly sorted
df = df.sort_values(by=['ID', 'trialseq'])

data, df = bebi103.stan.df_to_datadict_hier(
    df, level_cols="ID", data_cols=["trialseq", "DV", "US", "noUS"], 
    sort_cols=['ID', 'trialseq']
)

#pop returns the value and removes the key and value from the dictionary
data['participant_index'] = data.pop('index_1')
data['n_participants'] = data.pop('J_1')
data['n_stimuli'] = 1


sm = cmdstanpy.CmdStanModel(stan_file="Zbozinek_West_Perez_CompModeling.stan")

#Run MCMC
samples = sm.sample(data=data, chains=4, iter_sampling=1000, adapt_delta=0.995)
samples = az.from_cmdstanpy(
    posterior=samples, posterior_predictive="DV_ppc", log_likelihood="DV_log_like"
)

#Export Data
df_mcmc = bebi103.stan.arviz_to_dataframe(samples, df_package='pandas')
df_mcmc.to_csv(os.path.join(modeling_output_path,f'OS Transfer - Output.csv'))

metadata_df = df[['ID']].drop_duplicates().reset_index(drop=True)

# Extract alpha and attach IDs
alpha_medians = []
for col in df_mcmc.columns:
    if 'alpha_indiv' in col:
        alpha_medians.append(df_mcmc[col].mean())

alpha_medians = pd.DataFrame({
    'ext_alpha': alpha_medians,
    'ID': metadata_df['ID'].values
})

alpha_medians.to_csv(os.path.join(modeling_output_path, f'OS Transfer - Alpha Medians.csv'), index=False)


# Extract V0 and attach IDs
V0_medians = []
for col in df_mcmc.columns:
    if 'V0_indiv' in col:
        V0_medians.append(df_mcmc[col].mean())

# Convert to DataFrame with proper column names
V0_medians = pd.DataFrame({
    'ext_intercept': V0_medians,
    'ID': metadata_df['ID'].values  # Make sure order matches
})

# Save with named columns
V0_medians.to_csv(os.path.join(modeling_output_path, f'OS Transfer - V0 Medians.csv'), index=False)




# Plot Posterior

# Compute posterior predictive quantiles
qtiles = np.array(samples.posterior_predictive.DV_ppc.groupby('DV_ppc_dim_0').quantile([0.025, 0.5, 0.975], ...))
df[['low_ppc', 'median_ppc', 'high_ppc']] = qtiles

# Sort the dataframe
df_sorted = df.sort_values(['HierarchicalTraining', 'TransferTraining', 'ID'])

# Identify all unique condition combinations
condition_combos = df_sorted[['HierarchicalTraining', 'TransferTraining']].drop_duplicates().sort_values(['HierarchicalTraining', 'TransferTraining'])

# Map TransferTraining and HierarchicalTraining values to readable labels
transfer_map = {1: "Generalization", 2: "Ambiguous", 3: "Specificity"}
hierarchical_map = {0: "PR Control", 1: "True OS"}

# Store all condition-specific grids
for _, row in condition_combos.iterrows():
    hier = row['HierarchicalTraining']
    trans = row['TransferTraining']
    
    # Get the human-readable labels
    trans_title = transfer_map[trans]
    hier_title = hierarchical_map[hier]

    df_cond = df_sorted[(df_sorted['HierarchicalTraining'] == hier) & (df_sorted['TransferTraining'] == trans)]

    plots = []
    for subject, g in df_cond.groupby('ID'):
        p = bokeh.plotting.figure(
            frame_width=250,
            frame_height=150,
            toolbar_location=None,
            x_range=[0, 13],
            y_range=[df['low_ppc'].min(), df['high_ppc'].max()],
            x_axis_label='Extinction Trial',
            y_axis_label='US Expectancy (0-1)',
            title=f'{hier_title}, {trans_title}\nParticipant {subject}',  # Title with line break
        )
        p.yaxis.axis_label_text_font_style = 'normal'
        p.xaxis.axis_label_text_font_style = 'normal'

        x = np.concatenate((g['trialseq'].astype(int), np.array(g['trialseq'].astype(int))[::-1]))
        y = np.concatenate((g['low_ppc'], np.array(g['high_ppc'])[::-1]))
        p.patch(x, y, fill_alpha=0.3, fill_color='gray', line_alpha=0)
        p.line(g['trialseq'].astype(int), g['median_ppc'], line_width=2, line_color='gray')
        p.circle(g['trialseq'].astype(int), g['DV'])
        p.line(g['trialseq'].astype(int), g['DV'])

        plots.append(p)

    # Create a gridplot for this condition
    grid = gridplot(plots, ncols=3, toolbar_location=None)
    
    # Define the filename for this specific condition
    filename = f"Supplemental Figure 1 - {hier_title} {trans_title}.png"

    # Save the gridplot as a PNG file
    export_png(grid, filename=os.path.join(figures_path, filename))

    print(f"Saved plot for {hier_title}, {trans_title} as {filename}")
    bokeh.plotting.show(grid)
```

```python
# Figure 3

# Load and filter the data
df = pd.read_csv(os.path.join(stata_path, 'OS_Transfer_Exp_Ext_Alpha_V0.csv'))
df = df[df['TransferTraining'].isin([1, 2, 3])]
df = df[df['HierarchicalTraining'].isin([0, 1])]

# Plotting parameters
bar_width = 0.3
scatter_size = 4

# Outcome variable
outcome = 'ext_alpha'

p = bokeh.plotting.figure(
    frame_width=700,
    frame_height=400,
    x_axis_label='Transfer Training',
    y_axis_label='Alpha',
    x_range=(-0.25, 3.5),
    y_range=(-0.05, 1.05),
    toolbar_location=None,
    tools='',
)

for i, trans in enumerate([1, 2, 3], start=1):
    for ht in [0, 1]:
        group_df = df[
            (df['TransferTraining'] == trans)
            & (df['HierarchicalTraining'] == ht)
        ]

        if group_df.empty:
            continue

        # Offset bars to separate hierarchical training conditions
        x_offset = 0.175 if ht == 1 else -0.175
        x_val = i + x_offset

        sorted_df = group_df.sort_values(by=outcome).reset_index(drop=True)
        n = len(sorted_df)
        x_offsets = np.linspace(-bar_width / 2, bar_width / 2, n)

        # Individual participant observations
        for idx, row in sorted_df.iterrows():
            x = x_val + x_offsets[idx]
            y = row[outcome]

            p.scatter(
                x=[x],
                y=[y],
                size=scatter_size,
                alpha=0.7,
                color=Color4 if ht == 0 else Color8,
            )

        # Mean bar
        mean_val = group_df[outcome].mean()
        sem_val = group_df[outcome].sem()
        color = Color4 if ht == 0 else Color8
        legend_label = 'PR Control' if ht == 0 else 'True OS'

        p.vbar(
            x=x_val,
            top=mean_val,
            width=bar_width,
            color=color,
            alpha=0.4,
            legend_label=legend_label,
        )

        # SEM error bars
        cap = 0.05

        p.segment(
            x0=x_val,
            y0=mean_val - sem_val,
            x1=x_val,
            y1=mean_val + sem_val,
            line_color='gray',
            line_width=1,
            alpha=0.7,
        )

        p.segment(
            x0=x_val - cap,
            y0=mean_val + sem_val,
            x1=x_val + cap,
            y1=mean_val + sem_val,
            line_color='gray',
            line_width=1,
            alpha=0.7,
        )

        p.segment(
            x0=x_val - cap,
            y0=mean_val - sem_val,
            x1=x_val + cap,
            y1=mean_val - sem_val,
            line_color='gray',
            line_width=1,
            alpha=0.7,
        )


# --------------------------------------------------
# Add brackets and asterisks: PR Control vs True OS
# --------------------------------------------------
significance_stars = {
    1: '*',   # Generalization
    2: '**',  # Ambiguous
    3: '**',  # Specificity
}

bracket_y = {
    1: 0.96,
    2: 0.96,
    3: 0.96,
}

bracket_height = 0.025

for i, trans in enumerate([1, 2, 3], start=1):
    group_df = df[df['TransferTraining'] == trans]

    if group_df.empty:
        continue

    # X positions of PR Control and True OS bars
    x_control = i - 0.175
    x_true_os = i + 0.175

    y = bracket_y[trans]
    star = significance_stars[trans]

    # Horizontal bracket
    p.segment(
        x0=x_control,
        y0=y + bracket_height,
        x1=x_true_os,
        y1=y + bracket_height,
        line_color='black',
        line_width=1.5,
    )

    # Vertical bracket ends
    p.segment(
        x0=x_control,
        y0=y,
        x1=x_control,
        y1=y + bracket_height,
        line_color='black',
        line_width=1.5,
    )

    p.segment(
        x0=x_true_os,
        y0=y,
        x1=x_true_os,
        y1=y + bracket_height,
        line_color='black',
        line_width=1.5,
    )

    # Asterisks centered above the bracket
    p.text(
        x=[i],
        y=[y + bracket_height + 0.005],
        text=[star],
        text_align='center',
        text_baseline='bottom',
        text_font_size='14pt',
        text_font_style='bold',
        text_color='black',
    )

# Customize axes
p.xaxis.ticker = [1, 2, 3]
p.xaxis.major_label_overrides = {
    1: 'Generalization',
    2: 'Ambiguous',
    3: 'Specificity',
}

p.yaxis.ticker.desired_num_ticks = 11

# Font sizes
p.xaxis.axis_label_text_font_size = AxisLabelSize
p.xaxis.major_label_text_font_size = TickSize
p.xaxis.axis_label_text_font_style = 'normal'
p.yaxis.axis_label_text_font_size = AxisLabelSize
p.yaxis.major_label_text_font_size = TickSize
p.yaxis.axis_label_text_font_style = 'normal'

# Legend
p.legend.location = 'top_left'
p.legend.orientation = 'vertical'
p.legend.label_text_font_size = LegendSize

# Export alpha plot
export_png(
    p,
    filename=os.path.join(
        figures_path,
        'Figure 3.png',
    ),
)

# Display plot
bokeh.io.show(p)
```
